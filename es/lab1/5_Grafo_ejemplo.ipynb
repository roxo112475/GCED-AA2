```python
import torch

x = torch.tensor(2.5)
y = torch.tensor(3)

#print("El tensor x tiene datos del tipo: {0}\nEl tensor y tiene datos del tipo {1}\n".format(x.dtype, y.dtype))
#Podemos ver como las operaciones con tensores admiten numericos de diferentes tipos.

op1 = torch.add(x, y)        # suma
op2 = torch.multiply(x, y)   # multiplicación
op3 = torch.pow(op2, op1)    # potencia

print(f"El resultado de ({x.item()}*{y.item()})^({x.item()}+{y.item()}) es {op3.item()}")
```
